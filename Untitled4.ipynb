```python
import random

computer = random.randint(1, 10)
trial_limit = 0

guess =  int(input("Enter your guess: "))

while True:
  guess =  int(input("Enter your guess: "))
  trial_limit = trial_limit + 1

  if guess == computer:
    print("Correct guess")
    break
  elif trial_limit == 3:
    print("You lost. reached trial limit")
    break
  else:
    print("Wrong guess")
```

```python
import random

computer = random.choice(["rock", "paper", "scissors"])

while True:
  guess = input("Enter your guess: ")   # ❌ Not indented
  if guess.strip().lower() == computer:
    print("Correct guess")
    break
  else:
    print("Wrong guess")
```
